```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.optimizers import Adam

# 🧠 Load Pre-trained Emotion Model without compiling it
model = load_model('emotion_model.h5', compile=False)

# Recompile the model manually with correct arguments
model.compile(optimizer=Adam(learning_rate=0.0001), loss='categorical_crossentropy', metrics=['accuracy'])

# 🧠 Load Haar Cascade for face detection
face_detector = cv2.CascadeClassifier('haarcascades/haarcascade_frontalface_default.xml')

# 😄 Emotion Labels
emotion_labels = ['Anger', 'Disgust', 'Fear', 'Happy', 'Sad', 'Surprise', 'Neutral']

# 🧠 Start Video Capture from Webcam
cap = cv2.VideoCapture(0)

# Check if the webcam opened correctly
if not cap.isOpened():
    print("Error: Could not open the webcam.")
    exit()

# 😄 Process each frame
while True:
    ret, frame = cap.read()
    
    if not ret:
        print("Error: Could not read frame from the webcam.")
        break

    # 🧠 Convert to Grayscale for face detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 🧠 Detect faces
    faces = face_detector.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    for (x, y, w, h) in faces:
        # 🧠 Extract face from frame
        face = frame[y:y+h, x:x+w]
        
        # 🧠 Resize face to the input size for the model (64x64)
        face_resized = cv2.resize(face, (64, 64))
        
        # 🧠 Convert to grayscale (model expects grayscale input)
        face_resized = cv2.cvtColor(face_resized, cv2.COLOR_BGR2GRAY)

        # 🧠 Normalize the face image
        face_resized = face_resized.astype('float32') / 255.0
        face_resized = np.expand_dims(face_resized, axis=0)  # Add batch dimension
        face_resized = np.expand_dims(face_resized, axis=-1)  # Add channel dimension

        # 🧠 Predict emotion
        emotion_predictions = model.predict(face_resized)
        max_index = np.argmax(emotion_predictions[0])
        predicted_emotion = emotion_labels[max_index]

        # 🧠 Draw rectangle around face and label the emotion
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(frame, predicted_emotion, (x, y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    # 🧠 Display the resulting frame
    cv2.imshow("Emotion Detection", frame)

    # 🧠 Exit on pressing the 'q' key
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 🧠 Release resources and close all windows
cap.release()
cv2.destroyAllWindows()
```
